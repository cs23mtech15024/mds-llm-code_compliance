## Seed

```python
import json
from pathlib import Path

# Configuration
jsonl_file = r"/Users/sumandey/Desktop/IITH/Classes/Sem 5/mds-llm-code_compliance/Rules/5. Octal constants (other than zero) and octal escape sequences (other than '\0') shall not be used./Seeds/seed_1_100.jsonl"  # Change this to your JSONL file path
output_base_dir = r"/Users/sumandey/Desktop/IITH/Classes/Sem 5/mds-llm-code_compliance/Rules/5. Octal constants (other than zero) and octal escape sequences (other than '\0') shall not be used./Seeds"  # Note: capital P to match your folder name
start_program = 1  # Start from program number (not line number)
end_program = 100   # End at program number (not line number)

# First, let's verify the JSONL file exists
print(f"Looking for JSONL file: {jsonl_file}")
if not Path(jsonl_file).exists():
    print(f"ERROR: File '{jsonl_file}' not found!")
    print(f"Please make sure the file exists in the current directory.")
    exit(1)

print(f"✓ Found JSONL file")
print(f"Processing programs: {start_program} to {end_program}\n")

# Read and process the JSONL file
processed_count = 0
skipped_count = 0

with open(jsonl_file, 'r', encoding='utf-8') as f:
    for line_num, line in enumerate(f, start=1):
        # Skip empty lines
        if not line.strip():
            continue
        
        try:
            # Parse each JSON line
            data = json.loads(line.strip())
            
            # program_num = data['program']
            program_num = data['seed']

            # Skip programs outside our range
            if program_num < start_program or program_num > end_program:
                skipped_count += 1
                continue
            
            context = data['context']
            noncompliant_code = data['noncompliant']
            compliant_code = data['compliant']
            
            print(f"Processing program #{program_num} (from line {line_num})...")
            
            # Create folder name with zero-padding (051, 052, etc.)
            folder_name = f"{program_num:03d}"
            folder_path = Path(output_base_dir) / folder_name
            
            # Check if folder exists
            if not folder_path.exists():
                print(f"  ⚠ Warning: Folder {folder_name} does not exist, skipping...")
                continue
            
            # Create context comment
            context_comment = f"// Context: {context}\n\n"
            
            # Write mainNC.cpp (non-compliant)
            nc_file_path = folder_path / "seedNC.cpp"
            try:
                with open(nc_file_path, 'w', encoding='utf-8') as nc_file:
                    nc_file.write(context_comment)
                    nc_file.write(noncompliant_code)
                print(f"  ✓ Updated {folder_name}/seedNC.cpp")
            except Exception as e:
                print(f"  ✗ Error writing seedNC.cpp: {e}")
            
            # Write mainC.cpp (compliant)
            c_file_path = folder_path / "seedC.cpp"
            try:
                with open(c_file_path, 'w', encoding='utf-8') as c_file:
                    c_file.write(context_comment)
                    c_file.write(compliant_code)
                print(f"  ✓ Updated {folder_name}/seedC.cpp")
            except Exception as e:
                print(f"  ✗ Error writing seedC.cpp: {e}")
            
            processed_count += 1
            
        except json.JSONDecodeError as e:
            print(f"  ✗ Error parsing JSON at line {line_num}: {e}")
            continue
        except KeyError as e:
            print(f"  ✗ Missing key in JSON at line {line_num}: {e}")
            continue
        except Exception as e:
            print(f"  ✗ Unexpected error at line {line_num}: {e}")
            continue

print(f"\n{'='*50}")
print(f"✓ Complete!")
print(f"  Processed: {processed_count} programs")
print(f"  Skipped: {skipped_count} programs (outside range)")
print(f"  Range: programs {start_program} to {end_program}")
print(f"{'='*50}")
```
